# Collect CDF and time-series figures from all networks

Copies every `*_cdf.pdf` and `*_time_series.pdf` in `networks/<network>/latest_review/` into one OneDrive folder.

```python
from pathlib import Path
import shutil
from collections import Counter

# ---- Paths ----
SRC = Path("/workspaces/crmprtd/Quanlity_Control/variable_mapping/networks")
DEST = Path(
    "/workspaces/crmprtd/Quanlity_Control/variable_mapping/networks/figures")

# ---- Settings ----
PATTERNS = ["*_cdf.pdf", "*_time_series.pdf"]  # add more, e.g. "*_cdf.png"
ADD_NETWORK_PREFIX = False  # True -> network_1_net1_cdf.pdf
OVERWRITE = False           # True -> replace files that already exist in DEST
```

## Step 1: Preview what will be copied

```python
files = []
for net_dir in sorted(p for p in SRC.iterdir() if p.is_dir()):
    review = net_dir / "latest_review"
    if not review.is_dir():
        print(f"No latest_review folder in: {net_dir.name}")
        continue
    for pattern in PATTERNS:
        files.extend(sorted(review.glob(pattern)))

print(f"Found {len(files)} files\n")
for f in files:
    print(f.relative_to(SRC))

# Warn if different networks use the same file name (they would overwrite each other)
if not ADD_NETWORK_PREFIX:
    dupes = [n for n, c in Counter(f.name for f in files).items() if c > 1]
    if dupes:
        print("\nWARNING: duplicate file names, set ADD_NETWORK_PREFIX = True:", dupes)
```

## Step 2: Copy the files

```python
DEST.mkdir(parents=True, exist_ok=True)

copied, skipped = 0, 0
for f in files:
    name = f"{f.parent.parent.name}_{f.name}" if ADD_NETWORK_PREFIX else f.name
    target = DEST / name
    if target.exists() and not OVERWRITE:
        print(f"Skipped (already exists): {name}")
        skipped += 1
        continue
    shutil.copy2(f, target)
    print(f"Copied: {name}")
    copied += 1

print(f"\nDone. Copied {copied}, skipped {skipped}.")
print(f"Destination: {DEST}")
```
